# Exploratory Data Analysis (EDA), Lecture 04

```python
# import and setup
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('darkgrid')
```

# Fundamentals of Seaborn

```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

# Line Plot

```python
plt.figure(figsize=(20,4))
sns.lineplot(x='total_bill',y='tip',data=df)
plt.title('Total Bill vs Tip')
plt.savefig('lineplot.png')
plt.show()
```

# Bar Plot

```python
plt.figure(figsize=(15,4))
sns.barplot(x='day',y='total_bill',data=df)
plt.title('Day Wise Total Bill')
plt.savefig('barplot.png')
plt.show()
```

# Count Plot

```python
# Count plot counts the event
plt.figure(figsize=(15,4))
sns.countplot(x='smoker',data=df)
plt.title('Count Smokers')
plt.savefig('countplot.png')
plt.show()
```

# Box Plt

```python
plt.figure(figsize=(15,4))
sns.boxplot(x='day',y='total_bill',data=df)
plt.title('Day vs Total Bill')
plt.savefig('boxplot.png')
plt.show()
```

# Violin Plot

```python
plt.figure(figsize=(15,4))
sns.violinplot(x='day',y='total_bill',data=df)
plt.title('Day vs Total Bill')
plt.savefig('violin.png')
plt.show()
```

# Swarm Plot

```python
plt.figure(figsize=(15,4))
sns.swarmplot(x='day',y='total_bill',data=df)
plt.title('Day vs Total Bill')
plt.savefig('swarm.png')
plt.show()
```

# Strip Plot

```python
plt.figure(figsize=(15,4))
sns.stripplot(x='day',y='total_bill',data=df, jitter=False)   # if jitter=True, it acts like a swarm plot
plt.title('Day vs Total Bill')
plt.savefig('strip.png')
plt.show()
```

# Histogram

```python
plt.figure(figsize=(15,4))
sns.histplot(df['total_bill'], bins=30)
plt.title('Frequency count of Total Bill')
plt.savefig('histogram.png')
plt.show()
```

# KDE Plot

```python
plt.figure(figsize=(15,4))
sns.kdeplot(df['total_bill'],)
plt.title('Frequency Distribution line of Total Bill')
plt.savefig('kde.png')
plt.show()
```

# Histogram with KDE

```python
plt.figure(figsize=(15,4))
sns.histplot(df['total_bill'], bins=30, kde=True)
plt.title('Frequency count and Distribution Line of Total Bill')
plt.savefig('histogram with KDE.png')
plt.show()
```

# Heatmap

```python
corr = df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(15,4))
sns.heatmap(corr, annot=True)
plt.title('Heat Map of Correlation Matrix')
plt.savefig('HeatMap.png')
plt.show()
```

```python
sns.pairplot(df)
plt.title('Pair Plot')
plt.savefig('PairPlot.png')
plt.show()
```
